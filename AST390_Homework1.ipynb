```python
### Time for the imports!!

import numpy as np
import matplotlib.pyplot as plt
```

```python
### UUUUUUUUGGGGGHHHHHH??? ಥ_ಥ
### Date: 2/11/2025 
### Time: 5:31 PM 
### Maybe I did something wrong?
### It's time for me to redo my work again nonsensically.
### (ﾉ◕ヮ◕)ﾉ*:･ﾟ✧ Truly a definition of insanity trying to get the right answer.

# Suppress floating-point warnings
### I hope this works... ಠ╭╮ಠ
np.seterr(all='ignore')

def f(x):
  return np.sin(x)

def analytical_derivative(x):
    return np.cos(x)

# First-order one-sided finite difference
def one_sided_difference(f, x, dx):
    return (-3*f(x) + 4*f(x + dx) - f(x + 2*dx)) / (2 * dx)

# Second-order centered difference
def centered_difference(f, x, dx):
    return (f(x + dx) - f(x - dx)) / (2 * dx)

# Fourth-order accurate finite difference
def fourth_order_difference(f, x, dx):
    return (-f(x + 2*dx) + 8*f(x + dx) - 8*f(x - dx) + f(x - 2*dx)) / (12 * dx)
```

```python
### ᕦ(ò_óˇ)ᕤ Time to get serious!

# Define x value where we approximate the derivative
x = 1  
true_derivative = analytical_derivative(x)  # Exact derivative: cos(1)

# Generate a range of Δx values, avoiding extremely small values
dx_values = np.logspace(-8, 3, 12)  # From -8 to 2

# Compute errors for different methods
errors_one_sided = []
errors_centered = []
errors_fourth_order = []
```

```python
for dx in dx_values:
    errors_one_sided.append(abs(one_sided_difference(f, x, dx) - true_derivative))
    errors_centered.append(abs(centered_difference(f, x, dx) - true_derivative))
    errors_fourth_order.append(abs(fourth_order_difference(f, x, dx) - true_derivative))

# Compute convergence rates
def compute_convergence_rates(errors, dx_values):
    rates = []
    for i in range(len(dx_values) - 1):
        rate = np.log(errors[i+1] / errors[i]) / np.log(dx_values[i+1] / dx_values[i])
        rates.append(rate)
    return rates

rates_one_sided = compute_convergence_rates(errors_one_sided, dx_values)
rates_centered = compute_convergence_rates(errors_centered, dx_values)
rates_fourth_order = compute_convergence_rates(errors_fourth_order, dx_values)
```

```python
# Print the convergence rates
print("Δx\t\tOne-Sided Error\tCentered Error\t4th-Order Error")
print("="*70)
for i in range(len(dx_values) - 1):
    print(f"{dx_values[i]:8.6e}\t{errors_one_sided[i]:8.6e}\t{errors_centered[i]:8.6e}\t{errors_fourth_order[i]:8.6e}")
```

```python
# Plot error vs. dx for different methods
plt.figure(figsize=(8,6))
plt.loglog(dx_values, errors_one_sided, 'o-', label="One-Sided (1st Order)")
plt.loglog(dx_values, errors_centered, 's-', label="Centered (2nd Order)")
plt.loglog(dx_values, errors_fourth_order, 'd-', label="Fourth-Order")
plt.loglog(dx_values, dx_values**2, '--', label="Reference: O(Δx²)")
plt.loglog(dx_values, dx_values**4, '--', label="Reference: O(Δx⁴)")
plt.xlabel("Δx")
plt.ylabel("Absolute Error")
plt.legend()
plt.title("Comparison of Finite Difference Methods")
plt.grid(True)
plt.show()
```

```python
# Plot convergence rates
plt.figure(figsize=(8,6))
plt.plot(dx_values[:-1], rates_one_sided, 'o-', label="One-Sided (Expected: 1st Order)")
plt.plot(dx_values[:-1], rates_centered, 's-', label="Centered (Expected: 2nd Order)")
plt.plot(dx_values[:-1], rates_fourth_order, 'd-', label="Fourth-Order (Expected: 4th Order)")
plt.axhline(y=1, color='r', linestyle='--', label="O(Δx)")
plt.axhline(y=2, color='g', linestyle='--', label="O(Δx²)")
plt.axhline(y=4, color='b', linestyle='--', label="O(Δx⁴)")
plt.xscale("log")
plt.xlabel("Δx")
plt.ylabel("Convergence Rate")
plt.legend()
plt.title("Convergence Rate of Finite Difference Methods")
plt.grid(True)
plt.show()
```
